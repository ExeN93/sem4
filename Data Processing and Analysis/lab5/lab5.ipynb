# Selekcja zmiennych za pomocą przyrostu informacji

```python
import numpy as np
```

Zad 1

```python
def freq(x, prob=True):
    values = {}
    for value in x:
        if value in values:
            values[value] += 1
        else:
            values[value] = 1
    
    xi = np.array(list(values.keys()))
    ni = np.array(list(values.values()))
    
    if prob:
        ni = ni / ni.sum()
    
    return xi, ni
```
