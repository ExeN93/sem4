# Selekcja zmiennych za pomocą przyrostu informacji

```python
import numpy as np
import pandas as pd

from scipy import sparse
```

Zad 1

```python
x = [8,8,3,2,2,3,2,1,3,4,5,6,2,3,1,2,5]
def freq(x, prob=True):
    
    if sparse.issparse(x):
        x = x.toarray().flatten()
    
    values = {}
    for value in x:
        if value in values:
            values[value] += 1
        else:
            values[value] = 1
    
    xi = np.array(list(values.keys()))
    ni = np.array(list(values.values()))
    
    if prob:
        ni = ni / ni.sum()
    
    return xi, ni

print(freq(x, True))
print(freq(x, False))
```

Zad 2

```python
y = ['a', 'b', 'c', 'd', 'e', 'a', 'b', 'c', 'd', 'e']
x = [10, 20, 30, 40, 50, 10, 20, 30, 60, 90]

def freq2(x, y, prob=True):
    
    if sparse.issparse(x):
        x = x.toarray().flatten()
        
    if sparse.issparse(y):
        y = y.toarray().flatten()
    
    pairs = list(zip(x, y))
    unique_pairs, counts = np.unique(pairs, axis=0, return_counts=True)
    
    xi = np.unique(x)
    yi = np.unique(y)
    
    if prob:
        counts = counts / np.sum(counts)
    
    return xi, yi, counts

print(freq2(x, y, True))
print(freq2(x, y, False))
```

Zad 3

```python
x = [1, 5, 8, 5, 2]
y = ['a', 'a', 'a', 'b', 'b']

def entropy(x):
    values, probabilities = freq(x, prob=True)
    return -np.sum(probabilities * np.log2(probabilities))

def infogain(x, y):
    h_x = entropy(x)
    h_y = entropy(y)
    
    x_vals, y_vals, joint_probs = freq2(x, y, prob=True)
    h_xy = -np.sum(joint_probs * np.log2(joint_probs + 1e-10))
    
    return h_x + h_y - h_xy

print(f"H(X): {entropy(x)}")
print(f"H(Y): {infogain(x, y)}")
```

Zad 4

```python
plik = pd.read_csv("zoo.csv")
x = plik["type"]
plik = plik.drop(columns=["type", "animal"])
dane = []
for i in plik:
    y = plik[i]
    dane.append([i, entropy(y), infogain(x, y)])
    
df = pd.DataFrame(dane, columns=["columns", "entropy", "infogain"])
df = df.sort_values(by="infogain", ascending=False)

print(df)

dane2 = []

for i in range(0, plik.shape[1] - 1):
    for j in range (i+1, plik.shape[1]):
        dane2.append([plik.columns[i], plik.columns[j], infogain(plik[plik.columns[i]], plik[plik.columns[j]])])

df = pd.DataFrame(dane2, columns=["columns1", "columns2", "infogain"])
df = df.sort_values(by="infogain", ascending=False)

print(df)
```
